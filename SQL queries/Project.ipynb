# Retail Sales Analysis using PostgreSQL 


## Project Overview

This project focuses on analyzing retail sales data using PostgreSQL and visualizing business insights using Power BI for better business decision-making.


## Tools Used

- PostgreSQL
- SQL
- Power BI
- Visual Studio Code
- CSV Dataset


# Query 1 — Category-wise Total Sales


```python
SELECT 
    category,
    ROUND(SUM(sales), 2) AS total_sales
FROM superstore
GROUP BY category
ORDER BY total_sales DESC;
```

## Output Table

| Sl No | Category | Total Sales |
|---|---|---:|
| 1 | Office Supplies | 643707.69 |
| 2 | Technology | 470588.06 |
| 3 | Furniture | 451508.77 |


## Business Insight

Office Supplies generated the highest total sales among all categories, showing strong customer demand and consistent purchasing behavior.


# Query 2 — Region-wise Total Profit


```python
SELECT 
    region,
    ROUND(SUM(profit), 2) AS total_profit
FROM superstore
GROUP BY region
ORDER BY total_profit DESC;
```

## Output Table

| Sl No | Region | Total Profit |
|---|---|---:|
| 1 | West | 67859.96 |
| 2 | East | 53400.42 |
| 3 | Central | 27450.01 |
| 4 | South | 26551.72 |


## Business Insight

West region generated the highest total profit, indicating stronger business performance and better profitability compared to other regions.


# Query 3 — Top 5 Customers by Sales


```python
SELECT 
    customer_name,
    ROUND(SUM(sales), 2) AS total_sales
FROM superstore
GROUP BY customer_name
ORDER BY total_sales DESC
LIMIT 5;
```

# Query 4 — Monthly Sales Trend


```python
SELECT 
    TO_CHAR(TO_DATE(order_date, 'DD-MM-YYYY'), 'Mon-YYYY') AS month,
    ROUND(SUM(sales), 2) AS monthly_sales
FROM superstore
GROUP BY month
ORDER BY MIN(TO_DATE(order_date, 'DD-MM-YYYY'));
```

## Output Table

| Sl No | Month | Monthly Sales |
|---|---|---:|
| 1 | Jan-2019 | 18616.43 |
| 2 | Feb-2019 | 19978.81 |
| 3 | Mar-2019 | 51715.88 |
| 4 | Apr-2019 | 38750.04 |
| 5 | May-2019 | 50987.73 |


## Business Insight

Sales increased significantly during March and May, showing seasonal demand patterns and helping improve inventory planning.


# Query 5 — Loss-Making Products


```python
SELECT 
    product_name,
    ROUND(SUM(profit), 2) AS total_loss
FROM superstore
GROUP BY product_name
HAVING SUM(profit) < 0
ORDER BY total_loss ASC
LIMIT 10;
```

## Output Table

| Sl No | Product Name | Total Loss |
|---|---|---:|
| 1 | Cubify CubeX 3D Printer Double Head Print | -6239.98 |
| 2 | Cubify CubeX 3D Printer Triple Head Print | -3839.99 |
| 3 | Lexmark MX611dhe Monochrome Laser Printer | -2719.98 |
| 4 | Bush Advantage Collection Conference Table | -2019.24 |
| 5 | Ibico EPK-21 Electric Binding System | -1285.19 |


## Business Insight

Certain high-value products generated significant losses despite strong sales due to poor profit margins and pricing inefficiencies.


# Query 6 — Payment Mode Analysis


```python
SELECT 
    payment_mode,
    COUNT(*) AS total_orders
FROM superstore
GROUP BY payment_mode
ORDER BY total_orders DESC;
```

## Output Table

| Sl No | Payment Mode | Total Orders |
|---|---|---:|
| 1 | COD | 2453 |
| 2 | Online | 2164 |
| 3 | Cards | 1284 |


## Business Insight

COD was the most preferred payment method, followed by Online payments, showing customer preference for flexible payment options.


# Query 7 — Top Selling Products


```python
SELECT 
    product_name,
    ROUND(SUM(sales), 2) AS total_sales
FROM superstore
GROUP BY product_name
ORDER BY total_sales DESC
LIMIT 10;
```

## Output Table

## Business Insight

High-end office equipment and technology products contributed the highest sales revenue and play a major role in business growth.


# Query 8 — Worst Performing Products


```python
SELECT 
    product_name,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(SUM(profit), 2) AS total_profit
FROM superstore
GROUP BY product_name
ORDER BY total_profit ASC
LIMIT 10;
```

## Output Table

| Sl No | Product Name | Total Sales | Total Profit |
|---|---|---:|---:|
| 1 | Cubify CubeX 3D Printer Double Head Print | 9323.97 | -6239.98 |
| 2 | Cubify CubeX 3D Printer Triple Head Print | 3009.98 | -3839.99 |
| 3 | Lexmark MX611dhe Monochrome Laser Printer | 5676.97 | -2719.98 |
| 4 | Bush Advantage Collection Conference Table | 4334.10 | -2019.24 |
| 5 | Ibico EPK-21 Electric Binding System | 6437.97 | -1285.19 |


## Business Insight

Some products generated strong sales but still resulted in losses, highlighting the importance of analyzing both revenue and profitability together.


# Final Findings
